```python
import atlite
import matplotlib.pyplot as plt
import os

plt.style.use('seaborn-v0_8-darkgrid')
```

```python
# North Sea Coordinates
min_lon, max_lon = 5.0, 7.0
min_lat, max_lat = 54.0, 56.0
YEAR = 2023

cutout_path = f'../../data/NorthSea_{YEAR}.nc'
os.makedirs('../../data', exist_ok=True)

if not os.path.exists(cutout_path):
    cutout = atlite.Cutout(
        path=cutout_path,
        module='era5',
        x=slice(min_lon, max_lon),
        y=slice(min_lat, max_lat),
        time=slice(f'{YEAR}-01-01', f'{YEAR}-12-31')
    )
    cutout.prepare()
else:
    cutout = atlite.Cutout(cutout_path)
```

```python
# Calculate potential
offshore_wind = cutout.wind(turbine='NREL_ReferenceTurbine_15MW_offshore')

# Plot the spatial average over the year
plt.figure(figsize=(10, 5))
offshore_wind.mean(['x', 'y']).resample(time='1M').mean().plot(marker='o', color='teal')
plt.title('Monthly Average Offshore Wind Capacity Factor (North Sea)')
plt.ylabel('Capacity Factor')
plt.ylim(0, 1)  # Offshore often exceeds 0.5 capacity factor
plt.show()
```
