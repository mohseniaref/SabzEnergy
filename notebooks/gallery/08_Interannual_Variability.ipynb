```python
import matplotlib.pyplot as plt
from sabzenergy.climate import create_cutout
from sabzenergy.wind import calculate_wind_potential

plt.style.use('ggplot')
```

```python
CITY = 'Berlin'
YEAR_1 = 2022
YEAR_2 = 2023

cutout_1 = create_cutout(CITY, YEAR_1, output_dir='../../data', padding=0.5)
cutout_2 = create_cutout(CITY, YEAR_2, output_dir='../../data', padding=0.5)

wind_1 = calculate_wind_potential(cutout_1, turbine='Vestas_V112_3MW').mean(['x', 'y'])
wind_2 = calculate_wind_potential(cutout_2, turbine='Vestas_V112_3MW').mean(['x', 'y'])
```

```python
monthly_1 = wind_1.resample(time='1M').mean()
monthly_2 = wind_2.resample(time='1M').mean()

# Extract just the month values (1-12) to overlay them on the same axis
import pandas as pd
df = pd.DataFrame({
    '2022': monthly_1.values,
    '2023': monthly_2.values
}, index=range(1, 13))

df.plot(kind='bar', figsize=(10, 6), color=['#ff9999', '#66b3ff'])
plt.title(f'Inter-annual Wind Variability in {CITY}')
plt.xlabel('Month')
plt.ylabel('Average Capacity Factor')
plt.xticks(rotation=0)
plt.show()
```
