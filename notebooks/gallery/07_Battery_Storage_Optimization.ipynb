```python
'''
CITY = 'Munchen'
YEAR = 2023
cutout = create_cutout(CITY, YEAR, output_dir='../../data', padding=0.5)
pv_gen = calculate_solar_potential(cutout, panel='CSi').mean(['x', 'y']).to_series()

# Simple assumption: flat constant demand (e.g., 0.15 CF)
demand = 0.15
battery_capacity = 0.5  # Max capacity
battery_charge = 0.0

grid_feed = []
battery_level = []

for gen in pv_gen:
    if gen > demand:
        # Charge battery with excess
        excess = gen - demand
        charge_amount = min(excess, battery_capacity - battery_charge)
        battery_charge += charge_amount
        grid_feed.append(demand)  # Met demand
    else:
        # Discharge battery to meet demand
        deficit = demand - gen
        discharge_amount = min(deficit, battery_charge)
        battery_charge -= discharge_amount
        grid_feed.append(gen + discharge_amount)
        
    battery_level.append(battery_charge)
    
results = pd.DataFrame({'Generation': pv_gen, 'Grid_Feed': grid_feed, 'Battery_Level': battery_level}, index=pv_gen.index)
'''
```
